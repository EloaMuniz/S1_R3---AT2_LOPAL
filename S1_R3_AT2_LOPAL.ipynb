```python
#Exercício 2

lista = []

numero1 = float(input("Digite um número:"))
lista.append(numero1)

numero2 = float(input("Digite outro número:"))
lista.append(numero2)

numero3 = float(input("Digite o ultimo número:"))
lista.append(numero3)

print(f"A lista vale {lista}")

print(f"O maior número é {max(lista)}")

print(f"O menor número é {min(lista)}")

```

```python
#Exercício 5
nome = input("Digite seu nome: ")

while len(nome) <= 3:
    nome = input("Digite seu nome: ")

idade = int(input("Digite sua idade: "))

while idade < 0 or idade > 150:
    idade = int(input("Digite uma idade: "))

salario = float(input("Digite seu salário: "))

while salario <= 0:
    salario = float(input("Digite um salário: "))

sexo = input("Digite seu sexo (f/m): ")

while sexo != "f" and sexo != "m":
    sexo = input("Digite um sexo: ")

estado = input("Digite seu estado civil (s/c/v/d): ")

while estado != "s" and estado != "c" and estado != "v" and estado != "d":
    estado = input("Digite um estado: ")

print("Cadastro feito!👩‍💻")
```

```python
#Exercício 6

numero = int(input("Digite um número: "))

contador = 0

for i in range(1, numero + 1):

    if numero % i == 0:
        contador = contador + 1

if contador == 2:
    print("Número primo")

else:
    print("Número não primo")
```

```python
#Exercício 7

numero = int(input("Digite um número: "))

resultado = 1

for i in range(1, numero + 1):

    resultado = resultado * i

print(resultado)
```
